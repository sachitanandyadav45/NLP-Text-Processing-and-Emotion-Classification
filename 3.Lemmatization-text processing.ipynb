### Lemmatization in NLP
Lemmatization is an NLP technique used to convert words into their meaningful base or dictionary form, called a lemma.

Unlike stemming, lemmatization considers vocabulary and grammatical context to produce meaningful words.

```python
## Q & A,Chatbots,text summarization

import nltk

nltk.download('wordnet')
nltk.download('omw-1.4')
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer= WordNetLemmatizer()
```

```python
'''
POS:
Noun      - n
Verb      - v
Adjective - a
Adverb    - r
'''

lemmatizer.lemmatize("going", pos='v')
```

```python
words = ["eat", "eating", "eaten","history",
         "program", "programming", "programmer"]
```

```python
for word in words:
    print(word+"----->"+lemmatizer.lemmatize(word,pos='v'))
```
