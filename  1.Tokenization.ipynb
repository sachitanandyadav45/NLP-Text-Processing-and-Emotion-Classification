```python
corpus = """Natural Language Processing is an important branch of Artificial Intelligence.
It helps computers understand and process human languages.
Python provides several libraries for NLP, such as NLTK and spaCy.
Tokenization is one of the fundamental steps in NLP."""

print(corpus)
```

```python
import nltk
from nltk.tokenize import sent_tokenize

```

```python
documents=sent_tokenize(corpus)
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
### tokenization
## paragraph--->word
### sentence--->words
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer = TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```
