### Name entity recongnition

```python
"" Common NER Tags
NER Tag	Full Form	Example
PERSON	Person Name	Sachin, Ram
ORGANIZATION	Organization	Google, Microsoft
GPE	Geopolitical Entity	Nepal, India, Kathmandu
LOCATION	Location	Mount Everest, Himalayas
FACILITY	Building or Facility	Taj Mahal, Airport
DATE	Date	31 October, Monday
TIME	Time	10 AM, 5:30 PM
MONEY	Monetary Value	$500, Rs. 1000
PERCENT	Percentage	50%, 25%
CARDINAL	Cardinal Number	100, 500
ORDINAL	Ordinal Number	First, Second
EVENT	Event	Olympics, World Cup
PRODUCT	Product	iPhone, Windows
"""

```

```python
sentence="Sachin Yadav visited Kathmandu, Nepal, and attended a meeting at Google on 25 October 2026."
```

```python
import nltk
words=nltk.word_tokenize(sentence)
```

```python
tag_element=nltk.pos_tag(words)
```

```python
nltk.download('maxent_ne_chunker')
```

```python
nltk.ne_chunk(tag_element)
```
