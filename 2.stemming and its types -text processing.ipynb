### Stemming in NLP
Stemming is a text preprocessing technique in Natural Language Processing (NLP) that reduces words to their root or base form by removing prefixes or suffixes.

```python
## classification problem
## comments of product is a postive pr negative review
## review-->eating,eat,eaten[going,gone,goes]-->go
words = ["eat", "eating", "eaten","history",
         "program", "programming", "programmer"]
```

### porterstemmer

```python
from nltk.stem import PorterStemmer
```

```python
stemming=PorterStemmer()
```

```python
for word in words:
    print(word + " ----> " + stemming.stem(word))
    
```

```python
stemming.stem("congratulations")
```

### RegexpStemmer in NLTK
RegexpStemmer is a stemming technique in NLP that uses regular expressions (RegEx) to remove specified prefixes or suffixes from words.

```python
from nltk.stem import RegexpStemmer
```

```python

reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem('eating')
```

### Snowball Stemmer in NLTK
Snowball Stemmer is an NLP stemming algorithm used to reduce words to their root or stem form by removing prefixes and suffixes using predefined language-specific rules.

It is an improved version of the Porter Stemmer and supports multiple languages.

```python
from nltk.stem import SnowballStemmer
```

```python
snowballsstemmer= SnowballStemmer('english')
```

```python
for word in words:
    print(word+"---->"+snowballsstemmer.stem(word))
```

```python
stemming.stem("fairly"),stemming.stem("sportingly")
```

```python
snowballsstemmer.stem("fairly"),snowballsstemmer.stem("sportingly")
```
